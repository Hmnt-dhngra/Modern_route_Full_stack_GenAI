##### **Early Stage model training**

Early stage LLMs

**Here the parameters are of the order of a few millions**. 

These took less amount of data for unsupervised fine tuning and for supervised fine tuning they focussed on specific tasks like text classfification. 

We had 2 stages here:
- Pretraining : Unsupervised Pretraining / Self supervised learning
- Supervised fine tuning


1) Pretraining : Unsupervised Pretraining / Self supervised learning

In this stage, the model learns language patterns from a large amount of **unlabelled text data**. We provide the normal text data, the model automatically turns that into labelled data with the 4th word being the output and the first 3 words as input (basis window size of 3) and this process goes on in the shifting frame till the entire data gets exhausted. 

2 types : 
- Masked language modelling 
- **Auto regressive modelling** : This technique is being majorly used for pre-training these days. 


**Example: BERT**

BERT uses Masked language modelling to learn the language pattern. 

Input : The cat [MASK] on the mat
Target : sat

Here, the model learns to predict the missing word using both left and right context.



**Example: GPT**

GPT uses Auto regressive modelling (Next token prediction)

Input: I love Machine ______

Output: Learning

Here the model looks at previous tokens and predicts the next token. 

Example: 

Input : Artifical intelligence is 

Target: powerful

Then the sequence becomes : 

Artificial intelligence is powerful

Now GPT predicts the next token again

Input : Artificial intelligence is powerful
Target : because

Then the sequence becomes : 
Artificial intelligence is powerful because

So GPT learns by repeatedly predicting the next token from huge text data. 

In pretraining the model learns language structure, grammar, facts, and patterns without manually labelled data.

Further, the preference allignment is done as to what kind of the output is being prefered by human.

**ViT : Vision Transformer Explained**

Concept : 

Image → Patches → Patch embeddings → Transfomer encoder → Image representation

224 x 224 Image → 16 x 16 Patches → Patch tokens → Transfomer → Image representation

ViT treated images as patches, similar to words, and made transformers powerful in computer vision.

ViT paper showed that a pure Transformer can be applied directly to image patches and perform strongly on image recognition when trained at scale. 




**CLIP : Contrastive Language Image Pre-training Explained**

Concept : 

Image encoder + Text encoder 

Training : 

Correct image - text pair ko close lao
Wrong image - text pair ko door karo

Example : 

Image : Image of a [Dog running on a grass]
Text : A dog running on grass

CLIP image aur text ko same embedding space me map karta hai

Use cases : 
- Image search
- Zero-shot classification
- Text-guided image retrieval
- Multi-modal retrieval

CLIP brought images and text into a shared semantic space. It became a foundational block of multimodal AI.
