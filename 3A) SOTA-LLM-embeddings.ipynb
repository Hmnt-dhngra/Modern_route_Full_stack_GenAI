```python
from sentence_transformers import SentenceTransformer
```

```python
open_source_embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
## using this model, we can get the embedding for any sentence or paragraph or even a word. The embedding will be a vector of 384 dimensions.
```

#### **all-MiniLM-L6-v2 : This is supposed to be a compressed version of the BERT and is designed  to be smaller and faster while maintaining good performance for sentence embeddings.**

```python
text = "machine"
open_source_embedding_model.encode(text)
```

```python
para = "machine learning is fun."
para_embeddings= open_source_embedding_model.encode(para)
## same concept is used in Retrieval part of RAG systems
```

```python
len(para_embeddings)
# 384 is the dimensions of the embedding. 
```
