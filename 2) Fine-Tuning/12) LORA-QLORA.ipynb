There are 2 methods of fine-tuning
- Full Parameter Fine-tuning
- Parameter Efficient Fine-tuning 

#### **Full-parameter Fine-tuning**

- Updates all model weights
- The weight matrices are very large i.e. when we have a 7B model, it means it has 7 billion weights, 13B has 13 billion weights and so on..
- All weights get updated repeatedly for multiple epochs

*Storing and updatingw eights requires a lot of memory which limits fine-tuning to large GPUs or GPU clusters, hence this is very expensive*

#### **Parameter Efficient Fine-tuning (PEFT)**

- Does not update all the model weights
- Original weight matrices are frozen, only a small number of additional weights(parameters) are trained

##### **Ways to achieve PEFT**


- LoRa (Low-Rank Adaptation) --> QLoRA (Quantised Low-Rank Adaptation)
- DoRA (Weight-decomposed Low-Rank Adaptation)
- Prefix Tuning (adds learnable prefix vectors)
- Prompt Tuning (learns input embeddings)
- BitFit (trains only bias parameters)
- IA3 (scales attention mechanisms)


Most widely used is LoRA, DoRA is updated version of LoRA

Example : 
- LoRA trains small low-rank matrices instead of full weights. Training updates only these few parameters repeatedly for multiple epochs
- LoRA : Less memory + Faster training + Lower Cost + Easy Customisation, enabling fine-tuning on single GPUs or even consumer hardware. 
- Performance is almost same

#### **LoRA (Low-Rank Adaptation)**
- A method to achieve Parameter Efficient Finetuning

What is weights?

Weights = Importance score which shows which feature inside the neural network is important. 

The weights lie inside the layers, liuke we have linear layers that has weights like weights with Q, K, V which are called weight matrices. We also have attention weights which are MOST IMPORTANT, basis which the model decides which word is important for another word. Weights are represented as matrix. 

What happens during training ?

During the training (Forward propagation and backward propagation), the loss is calculated and weights are updated using gradients which were initially randomly allocated. The goal is to get the best weights values that minimizes the loss.
