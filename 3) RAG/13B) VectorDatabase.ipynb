### **Vector Database**

#### **What is Vector Database**

A vector database stores data as numerical vectors called embeddings and retrieves the most similar vectors using similarity search. 

It is mainly used for semantic search, RAG, recommendation systems, image search where results are found by meaning rather than exact keyword matching. 

#### **What is Vector Index**

A vector index is a data structure that organises embeddings so the vector database can find similar vectors quickly without comparing the query with every stored vector. 

**Example :** 
- Flat
- IVF
- HNSW

Vector database stores vectors, the vector index makes searching those vectors faster. 

#### **What is Similarity Search**

Similarity search is the process of finding stored vectors that are closest or most similar to a query vector. 

It compares vectors using metrics such as cosine similarity, Euclidean distance or dot product. 

**Example:**

A search for **"How can I recover my account?"** may retrieve **"Steps to reset your password"** because both have similar meaning, even though the words are entirely different. 

Vector Similarity search can be categorised as follows : 
- **Exact nearest neighbour search** : It checks the query againts every stored vector and returns the true nearest matches. 
    - It is 100% exact as it works on the method of exact search.
    - Provides accuracy
    - It is slower on large datasets
    - Suggestible for small datasets or evaluation
    - Examples : Flat index
- **Approximate nearest neighbour search (ANN)** : It searches only a selected part of the index to return very close matches faster, but it may occasionally miss the exact best result. 
    - Approximate search, high recall
    - Prioritizes speed and scalability. 
    - Much faster
    - Suggestible for large-scale production search. 
    - Example : HNSW, IVF

#### **Vector Databases can be categorised as follows basis the vector management**

- **In-memory** : FAISS, ChromeDB
- **Local persistent** : FAISS, ChromaDB, Qdrant, Milvus
- **Vendor-managed SaaS (Vector DB)** : The database vendor managed it. Pinecone, Qdrant, Milvus, Weaviate
- **Cloud-provider managed** : AWS, Azure or GCP manages it. Azure AI Search, AWS OpenSearch, Google's Vertex AI vector search, Google's BigQuery vector search, etc.
