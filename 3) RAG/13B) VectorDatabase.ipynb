### **Vector Database**

#### **What is Vector Database**

A vector database stores data as numerical vectors called embeddings and retrieves the most similar vectors using similarity search. 

It is mainly used for semantic search, RAG, recommendation systems, image search where results are found by meaning rather than exact keyword matching. 

#### **What is Vector Index**

A vector index is a data structure that organises embeddings so the vector database can find similar vectors quickly without comparing the query with every stored vector. 

**Example :** 
- Flat
- IVF (Inverted File Index)
- HNSW (Hierarchical Navigable Small world)

Vector database stores vectors, the vector index makes searching those vectors faster. 

#### **What is Similarity Search**

Similarity search is the process of finding stored vectors that are closest or most similar to a query vector. 

It compares vectors using metrics such as cosine similarity, Euclidean distance or dot product. 

**Example:**

A search for **"How can I recover my account?"** may retrieve **"Steps to reset your password"** because both have similar meaning, even though the words are entirely different. 

Vector Similarity search can be categorised as follows : 
- **Exact nearest neighbour search** : It checks the query againts every stored vector and returns the true nearest matches. 
    - It is 100% exact as it works on the method of exact search.
    - Provides accuracy
    - It is slower on large datasets
    - Suggestible for small datasets or evaluation
    - Examples : Flat index
- **Approximate nearest neighbour search (ANN)** : It doesn't iterate on every row. It searches only a selected part of the index to return very close matches faster, but it may occasionally miss the exact best result. 
    - Approximate search, high recall
    - Prioritizes speed and scalability. 
    - Much faster
    - Suggestible for large-scale production search. 
    - Example : HNSW (uses graph based approach for approximation), IVF (uses cluster based approach for approximation)

**Flat Index Exact match**

Suppose we store 1 million records inside the flat index. Then you ask for similarity search of a user query, now that query will interact with all the 1 million records and will perform linear search. Hence it is time consuming. 

**IVF Index Approximate Cluster Based approach**

Suppose we store 1 million records inside the IVF index. It will cluster the input records into multiple clusters (arrangemnent of data). When you ask for similarity search of a user query, now that query will interact with all the clusters and the cluster having the least distance will be tried for every records inside that. 


**HNSW Index Approximate Graph based approach**

Suppose the arrangement is like a graph, it follows this structure. Whenever a query is asked, it follows a specific path and based on that it will try to find the relevant results instead of going into every row of the record.

#### **Vector Databases can be categorised as follows basis the vector management**

- **In-memory** : FAISS, ChromeDB
- **Local persistent** : FAISS, ChromaDB, Qdrant, Milvus
- **Vendor-managed SaaS (Vector DB)** : The database vendor managed it. Pinecone, Qdrant, Milvus, Weaviate
- **Cloud-provider managed** : AWS, Azure or GCP manages it. Azure AI Search, AWS OpenSearch, Google's Vertex AI vector search, Google's BigQuery vector search, etc. 

#### **What all is stored inside Vector Databases**

##### **The Vector Database Storage Pipeline**

Raw Data (Text/Image/Audio/Video)  
    ↓     
Embedding Model (Convert to numerical vectors)  
    ↓  
Vector Representation (e.g., 768-dimensional array)  
    ↓  
Metadata Association (Link original data + context)  
    ↓  
Indexing Strategy (Organize vectors for fast search)  
    ↓  
Storage Backend (Persist vectors + metadata)  
    ↓  
Vector Database (Complete system ready for querying)

Data (Text/Image/Audio/Video) => Vectors => Indexed

Indexes can be categorised basis exact match (Flat index) and approximate match (IVF Index or HNSW Index). 

Inside the index itself we store the data. 

The data stored can constitute 
- Actual data
- Metadata
- Embeddings corresponding to actual data
