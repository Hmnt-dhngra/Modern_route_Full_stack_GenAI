#### **Data Parsing using Docling**

Open source library for data parsing i.e. data extraction and arrangement into jason, md, html, etc. formats. This can be downloaded locally. 

Docling is very capable

We have more examples of data parsing like llamaparsr. This can't be downloaded locally but can be accessed via llama-cloud. 

The first part before starting the data extraction depends on the kind of PDF we have. We will use the data parsing technique basis what kind of data the PDF holds. 

```python
import json
from pathlib import Path

from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import (
    PdfPipelineOptions,
    TableStructureOptions,
    TesseractCliOcrOptions,
)
from docling.document_converter import (
    DocumentConverter,
    PdfFormatOption,
)
from docling_core.types.doc import (
    ImageRefMode,
    PictureItem,
    TableItem,
)
```

```python
PDF_PATH = Path(".\RAG_data\complex_rag_parsing_sample_with_sunny_image.pdf")

OUTPUT_DIR = Path(".\RAG_data\docling_parsed_output")

PAGE_IMAGE_DIR = OUTPUT_DIR / "page_images"
PICTURE_DIR = OUTPUT_DIR / "extracted_pictures"
TABLE_DIR = OUTPUT_DIR / "extracted_tables"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PAGE_IMAGE_DIR.mkdir(parents=True, exist_ok=True)
PICTURE_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)

if not PDF_PATH.exists():
    raise FileNotFoundError(f"PDF not found: {PDF_PATH}")
```
