```python
from langchain_community.document_loaders import (
    TextLoader, 
    PyPDFLoader,
    CSVLoader, 
    Docx2txtLoader,
    UnstructuredHTMLLoader,
    JSONLoader)
```

In Langchain, we call them document loaders. They load files like PDF, DOCX, CSV, TXT, HTML and JSON. Internally they may use parser libraries like pypdf, docx2txt, unstructured, or jq to extract the content. Then LangChain converts that extracted content into Document format(page_content, metadata) for RAG pipeline. 

In LangChain, parsed content comes in the form of a Document object. 

Inside the object :

- Page_content = the actual extracted text
- metadata= source information like file name, page number, row number, etc.
