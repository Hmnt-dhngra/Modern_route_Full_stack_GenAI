## **Introduction to Agentic AI**

### **1) What is an AI Assistant?**

An **AI assistant** is a software system powered by AI that generates output of user's questions basis the model's available knowledge (already available training data). 

It typically:
- Receives an instruction from the user.
- Processes the input using an AI model.
- Generates a relevant response.

Ques : What is Tesla

Answer : Tesla is an EV company. 

This answer is generated basis the model's default knowledge

### **2) What is a RAG Based Assistant?**

A **RAG (Retrieval-Augmented Generation) assistant** is an AI assistant that retrieves relevant external knowledge and provides that information as context to the AI model to generate a response.

It typically:
- Receives a question from the user.
- Searches a knowledge base (Vector Database) for relevant information.
- Adds the retrieved information to the prompt.
- Generates an answer based on the retrieved context.

**Example:**  
If a user asks about a company's internal policy, a RAG assistant retrieves the relevant policy document and answers using that document rather than relying only on the model's training data.

RAG improves response accuracy, supports up-to-date or private information, and helps reduce hallucinations.


### **3) What is an agent?**

An **agent** is an autonomous system or simply said an intelligent system. An agent can be called as an AI system that can perceive its environment, reason about a goal, make decisions, and take actions using available tools.

User Input (Goal) --> Think/Reasoning --> Planning --> Choose action (Tool calling) --> Observe/check the results --> Repeating the loop until not getting the final answer

Ths entire system of doing these things is called an agent where the thinking ability is coming from the LLM. 

An AI agent is an AI system that can autonomously decide and execute actions using tools and environment feedback to achieve a goal. 

Example : User says : Analyse Tesla and prepare an investement research report

Agent :
I need current financial information --> Use web search --> I need hsitorical stock data --> Use Finance API --> I need to do calculations --> Use Python --> I need Tesla's annual report --> Search Docs --> Do i have enough information --> No -> Search again Yes -> Generate Report


### **4) What is a multi-agentic system?**

A multi agent system uses multiple agents instead of one agent. Each agent can specialise in a particular job. 

We may have one supervisor Agent (which may think/act/observe/do an action)

Helper Agents
- Research Agent (think/act/observe)
- Finance Agent (think/act/observe)
- Risk Agent (think/act/observe)

**Example**
Tesla Example 
User : Analyse Tesla and prepare a detailed investment research report

A supervisor may delegate :
1) Research Agent for research about Tesla products, market, competitors, Recent developments, etc.
2) Financial Agent analyses revenue, profit, cash flows, margins, etc.
3) Risk Agent analyses regulatory risks, competition, supply-chain risks
4) Market Agent analyses EV market, competitors, etc.
5) Writer Agent combines everything into one report. 

So instead of 

One agent who does everything we have multiple specialised agents who collaborate and delegate together and complete the goal. 

### **5) What is a Deep Agentic System?**

It is a large and long running system which are usually used to solve a complex problem. Here we will have extensive planning. 

Complex Goal --> Extensive Planning --> Task decomposition --> Multiple tasks --> Sub-agents --> Repeat --> Final deliverable. 

- Single/monolithic agentic systems
- Multi-agentic systems 

These both could be deep agentic systems involving deep extensive planning involving multiple steps for long tasks execution for complex problems. 

## **Flow of an agentic system**

Input --> Think/reasoning/Decision --> Planning if necessary --> Action --> Observe the results --> Adapt based on feedback(repeat the steps if needed) i.e. Iterative Refinement --> Output

This flow can be divided into 2 parts i.e. 
1) **Deterministic Flow**
2) **Dynamic Flow**
