## **Flow of an agentic system**

Input --> Think/reasoning/Decision --> Planning if necessary --> Action --> Observe the results --> Adapt based on feedback(repeat the steps if needed) i.e. Iterative Refinement --> Output

This flow can be divided into 2 parts i.e. 
1) **Deterministic Flow**
2) **Dynamic Flow**

This flow can be implemented using 
1) Pure Python or
2) Langgraph or ***Winner***
3) OpenAI Agent SDK  or ***2nd Winner***
4) Google ADK or 
5) Autogen or
6) CrewAI or
7) SmolAgent (HuggingFace)
