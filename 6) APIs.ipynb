##### **LangChain**

All the LLM providers mentioned above have their own SDKs (Software development kits). Using each SDK individually you can directly call individual models separately using separate SDKs. 
Langchain is integrating all these into its own SDK and acting as a unified framework for all the APIs SDK so that using one single LangChain SDK u can call any of the models. 

LangChain and individual providers SDK are not limited to model calling, they have a wider purpose that involves multiple components like :
- Prompt Templates
- Memory
- Document Loaders
- Text Splitters
- Embeddings
- Vector DBs
- Retrievers
- Tools
- Agents
- Tracing
- Evaluation
- Model Switching

Managing all these for separate LLm providers would be cumbersome, hence LangChain came into picture which provides a unified framework to be able to work with multiple LLM providers and their components. 


LangChain is a US based company that has provided open source frameworks like 
- LangChain : Opensource
- LangGraph : Opensource
- LangSmith : Paid Platform, Enterprises buy this and is mainly used for Evaluation, traceability and Observability
- Langfuse : Platform Observability
